```python
import pandas as pd 
import numpy as np
```

```python
movies = pd.read_csv("C:/Users/Apratim Dey Tapadar/OneDrive/Desktop/PROJECT/Movie_recommendation_system/data/tmdb_5000_movies.csv")
credits = pd.read_csv("C:/Users/Apratim Dey Tapadar/OneDrive/Desktop/PROJECT/Movie_recommendation_system/data/tmdb_5000_credits.csv")
```

```python
movies=movies.merge(credits, on = 'title')
```

```python
movies=movies[["movie_id","title",
        "overview",
        "genres",
        "keywords",
        "cast",
        "crew"
]]
```

```python
movies.shape
```

```python
pd.set_option("display.max_columns", None)
```

```python
movies.head()
```

```python
movies.dropna(inplace = True)
```

```python
movies.isnull().sum()
```

```python
import ast 
```

```python
def convert(text):
    L=[]

    for i in ast.literal_eval(text):
        L.append(i["name"])

    return L
    
```

```python
movies["genres"]=movies["genres"].apply(convert)
```

```python
movies["genres"].head()
```

```python
movies["keywords"] = movies["keywords"].apply(convert)
```

```python
movies["keywords"].head()
```

```python
movies.head()
```

```python
def convert_cast(text):
    L = []
    counter = 0

    for i in ast.literal_eval(text):

        if counter != 3:
            L.append(i["name"])
            counter += 1

        else:
            break

    return L

```

```python
movies["cast"] = movies["cast"].apply(convert_cast)
```

```python
def fetch_director(text):
    L=[]

    for i in ast.literal_eval(text):

            if i["job"] == "Director":
                L.append(i["name"])
                break

    return L
```

```python
movies["crew"] = movies["crew"].apply(fetch_director)
```

```python
movies.head()
movies.shape
movies.info()
movies.isnull().sum()
```

```python
movies["overview"] = movies["overview"].apply(lambda x: x.split())
```

```python
movies["overview"].head()
```

```python
def collapse(L):
    L1 = []

    for i in L:
        L1.append(i.replace(" ", ""))

    return L1
```

```python
movies["genres"] = movies["genres"].apply(collapse)

movies["keywords"] = movies["keywords"].apply(collapse)

movies["cast"] = movies["cast"].apply(collapse)

movies["crew"] = movies["crew"].apply(collapse)
```

```python
movies["tags"] = (
    movies["overview"] +
    movies["genres"] +
    movies["keywords"] +
    movies["cast"] +
    movies["crew"]
)
```

```python
movies["tags"].head()
```

```python
df = movies[["movie_id", "title", "tags"]]
```

```python
df.head()
```

```python
df["tags"] = df["tags"].apply(lambda x: " ".join(x))
```

```python
df["tags"] = df["tags"].apply(lambda x: x.lower())
```

```python
df["tags"].head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(
    max_features = 5000,
    stop_words = "english"
)
```

```python
vectors = cv.fit_transform(df["tags"]).toarray()
```

```python
vectors.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
print(similarity)
```

```python
def recommend(movie):

    movie_index = df[
        df["title"] == movie
    ].index[0]

    distances = similarity[movie_index]

    movies_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]

    for i in movies_list:
        print(df.iloc[i[0]].title)
```

```python
recommend("Batman Begins")
```

```python
recommend("Avatar")
```
